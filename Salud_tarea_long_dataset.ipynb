```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Salud_Mental_20260412.csv") #Cargamos el archivo
df.head() #Visualizamos la estructura del dataframe (detectamos arreglo wide)
```

```python
df.info() #Revisamos la informacion del dataset
```

```python
print(df.isnull().sum()) #Tenemos todas las casillas llenas, por lo que no tenemos necesidad de hacer un tratamiento de datos nulos.
print(df.columns) #revisamos que columnas tenemos
```

```python
df['diagnostico_ingreso'] #Revisamos que enfermedades tenemos
```

```python
df_cols_edades = ['De 1 a 4','De 5 a 9','De 10 a 14', 'De 15 a 19','De 20 a 49','De 50 a 64'] #definimos las columnas que nos interesan para la tarea
#El dataset está en formato wide(ancho), lo necesitamos en formato long.
df_long = df.melt( #Aplicamos la funcion melt para convertir nuestro dataset en formato long
    id_vars='diagnostico_ingreso', #id_vars ahora seran las enfermedades (diagnostico_ingreso)
    value_vars=df_cols_edades, #El value vars seran las columnas que previamente definimos, en este caso, los rangos de edad
    var_name='Edad_Rango', #Definimos el nombre de la segunda columna
    value_name='Casos' #Definimos el nombre de la 3er columna que será la cuenta de repetición de id_vars en Edad_Rango
)
df_long.loc[df_long.groupby('Edad_Rango')['Casos'].idxmax()] #Ya con el dataset en long, hacemos un groupby de Edad rango y casos, encontramos
# segun el diagnostico_ingreso los valores mas altos y los listamos. asi obtenemos las enfermedades que mas se repiten segun cada rango de edad.
```
