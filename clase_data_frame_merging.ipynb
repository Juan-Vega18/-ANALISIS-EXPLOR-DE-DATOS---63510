```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
movies = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movies.csv')
ratings = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/ratings.csv')
```

```python
print(f'El head de movies es: \n\n{movies.head(30)}')
print('*'*60)
print(f'El info de movies es: \n\n{movies.info}')
print('*'*60)
print(f'El shape de movies es: \n\n{movies.shape}')
```

```python
print(f'El head de ratings es: \n\n{ratings.head(30)}')
print('*'*60)
print(f'El info de ratings es: \n\n{ratings.info}')
print('*'*60)
print(f'El shape de ratings es: \n\n{ratings.shape}')
```

La metodologia consiste en:

`df1.merge(df2, aqui hay que darle la columna con la cual vamos a unir, esperamos que se llamen igual)`

por ejemplo, en nuestro caso puede ser `id`
`df1.merge(df2, on='id')`

```python
#Unir los dataframes
movie_ratings = movies.merge(ratings, on='id')
movie_ratings.drop(['Unnamed: 0_x','Unnamed: 0_y'], axis=1, inplace=True)
movie_ratings.head()

#axis = 1 significa columnas
#inplace = True, modifica el dataframe original y lo reemplaza.
```

```python
movie_genders = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movie_to_genres.csv')
movie_genders
```

```python
movie_genders.drop(['Unnamed: 0'], axis=1, inplace=True)
```

```python
movie_genders
```

```python
movies_with_genders = movies.merge(movie_genders, left_on='id', right_on='movie_id')
#left on es el dataset fuera de merge
#el right on es el dataset dentro de merge.
```

```python
movies_with_genders
```

```python
movies_with_genders_copy = movies_with_genders.copy()
print(movies_with_genders_copy['popularity'].sort_values())
print(movies_with_genders_copy.iloc[2107])
```

```python
sci_movies = movies_with_genders[movies_with_genders['genre'] == 'Science Fiction']
sci_movies
#sci_movies.shape
```

```python
action_movies = movies_with_genders[movies_with_genders['genre'] == 'Action']
action_movies
```

```python
df_left = action_movies.merge(sci_movies, on='movie_id', how ='left')
```

```python
df_left
```

```python
df_left_suf = action_movies.merge(sci_movies, on='movie_id', how ='left', suffixes=("_action","_scifi"))
```

```python
df_left_suf
```

```python
#Ahora una union derecha
df_r_suf = action_movies.merge(sci_movies, on='movie_id', how ='right', suffixes=("_action","_scifi"))
```
