Aqui nos damos cuenta de varias cosas y trucos del profe jeje (del dataset realmente). Resulta que tenemos unas fechas en diferente formato, y algunas otras imposibles.

Por ejemplo:
`time data "2021/02/30" doesn't match format "%Y-%m-%d %H:%M:%S", at position 100`

Esto literalmente es imposible ya que febrero no tiene dia 30...
Aparte el formato correcto es con `-` no con `/`, y algunos datos estan mixed. Asi que necesitamos hacer una limpieza mas profunda.

```python
df['fecha_consulta'] = pd.to_datetime(
    df['fecha_consulta'],
    errors='coerce',
    format='mixed'
)
```

¿Que hacen los dos parametros que fijamos?
1. `errors='coerce'` Lo que hace es convertir las fechas sin sentido como `2021/02/30` en `NaT` que significa `Not a Time`.
2. `format='mixed'` lo que hace es permitir multiples formatos, con guiones, `/`, incluso con las horas (nuestro caso)

```python
#1 Debemos separar la presion en sistolica y diastolica
#en dos nuevas columnas

df[['presion_sistolica', 'presion_diastolica']] =(
    df['presion_arterial']
    .astype(str)
    .str.strip()
    .str.split('/', expand=True)
)

#Convertir todo en numerico, lo que no lo sea, de una a nan
df['presion_sistolica'] = pd.to_numeric(
    df['presion_sistolica'],
    errors='coerce'
)

df['presion_diastolica'] = pd.to_numeric(
    df['presion_diastolica'],
    errors='coerce'
)
```

```python
df.head()
```

Reglas que vamos a poner
Sistolica razonable (segun google, yo no se de esto)
Minima plausible viva: aprox 70
Maxima plausible extrema: aprox 300

Diastolica razonable:
minima plausible: aprox 40
maxima extrema: aprox 180

```python
df['presion_sistolica'] = df['presion_sistolica'].where(
    (df['presion_sistolica'] >= 70) &
    (df['presion_sistolica'] <= 300),
    np.nan
)

df['presion_diastolica'] = df['presion_diastolica'].where(
    (df['presion_diastolica'] >= 40) &
    (df['presion_diastolica'] <= 180),
    np.nan
)
```

```python
#La sistólica debe ser mayor a la diastolica (google dice esto)
#por lo tanto añadimos una nueva condicion

# sistolica > diastolica
mask = df['presion_sistolica'] <= df['presion_diastolica']

#Lo que no cumpla la mascara que pusimos automaticamente se va a NaN
df.loc[mask, ['presion_sistolica', 'presion_diastolica']] = np.nan
```

```python
#Ahora vamos con la glucosa.
#Ahora glucosa. Aplicamos el mismo estudio para la glucosa
print(df['glucosa_mg_dL'].dtype)
print('#'*40)
print(df['glucosa_mg_dL'].isnull().sum())
print('#'*40)
print(df['glucosa_mg_dL'].duplicated())
print('#'*40)
print(df['glucosa_mg_dL'].sample())
print('#'*40)
print(df['glucosa_mg_dL'].head(30))
print('#'*40)
df_copy['glucosa_mg_dL'].astype(str).str.strip().value_counts().head(30)
```

La columna de `glucosa_mg_dL` esta fuertemente contaminada con un monton de errores, por ejemplo, cuando hacemos el value_counts, nos encontramos que hay valores negativos, valores como (alto, bajo) que no nos sirven como caracterisitca numerica, o NaN, tambien valores muy altos como:
300 (extremo pero posible en crisis)
500 (muy extremo, raro pero médicamente posible)

```python
glucosa = (
    df['glucosa_mg_dL']
    .astype(str)
    .str.strip()
    .str.lower()
)
glucosa = glucosa.replace({
    'alto': np.nan,
    'bajo': np.nan,
    'nan': np.nan
})
glucosa = pd.to_numeric(glucosa, errors='coerce')

glucosa = glucosa.where(
    (glucosa >= 20) & (glucosa <= 600),
    np.nan
)
#<20 casi imposible con una persona viva
#>600 muy raro, lo tomaremos como error
df['glucosa_mg_dL'] = glucosa

```

Por ultimo vamos a calcular el imc con todo el dataset limpio... luego, borraremos las filas vacias con alguna estrategia.
IMC = peso(kg)/altura^2 (m)

```python
df['imc_calculado'] = np.where(
    (df['peso_kg'].notna()) & #los no vacios
    (df['altura_m'].notna()), #los no vacios
    df['peso_kg'] / (df['altura_m'] ** 2),
    np.nan
).round(2)
```
